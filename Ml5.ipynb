```python

import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
data = pd.read_csv("/content/car data.csv")
data
```

```python
data.shape
```

```python
data.isnull()
```

```python
data.isnull().sum()
```

```python
data.duplicated().sum()
```

```python
data = data.drop_duplicates()
data
```

```python
print("\n Duplicate values")
print(data.duplicated().sum())
```

```python
le = LabelEncoder()
```

```python
data.loc[:, "Fuel_Type"] = le.fit_transform(data["Fuel_Type"])
data.loc[:, "Seller_Type"] = le.fit_transform(data["Seller_Type"])
data.loc[:, "Transmission"] = le.fit_transform(data["Transmission"])
```

```python
print(data.head())
```

```python
X = data[["Year","Present_Price","Kms_Driven","Fuel_Type","Seller_Type","Transmission","Owner"]]
Y = data["Selling_Price"]
print(X)
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
model = LinearRegression()
model.fit(X_train, Y_train)
```

```python
Y_pred = model.predict(X_test)
```

```python
print("Predicted Selling Prices:")
print(Y_pred)
```

```python
mae = mean_absolute_error(Y_test, Y_pred)
mse = mean_squared_error(Y_test, Y_pred)
r2 = r2_score(Y_test, Y_pred)
```

```python
print("Mean_Absolute Error:", mae)
print("Mean_Squared Error:", mse)
print("R2 Score:", r2)
```
